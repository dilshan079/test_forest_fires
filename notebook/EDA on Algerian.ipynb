ഫീച്ചറുകൾ തമ്മിൽ അമിതമായി കോറിലേഷൻ ഉള്ള അവസ്ഥയെ Multicollinearity എന്ന് വിളിക്കുന്നു

```python
def correlation(X_train, threshold):
    col_corr = set()

    
    corr_matrix = X_train.corr()

    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i, j]) > threshold:
                colname = corr_matrix.columns[i]
                col_corr.add(colname)

    return col_corr
```

```python
core_features = correlation(X_train,0.85)
```

```python
X_train.drop(core_features,axis=1,inplace=True)
X_test.drop(core_features,axis=1,inplace=True)
X_train.shape,X_test.shape
```

 ##Standardization of the data

```python

from sklearn.preprocessing import StandardScaler    

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
#Box plot to understand the effect of standardization on the features
```

```python
plt.subplots(figsize=(20,10))
plt.subplot(1, 2, 1)
sns.boxplot(data=X_train)
plt.title("Before Standardization")
plt.subplot(1, 2, 2)
sns.boxplot(data=X_train_scaled)
plt.title("After Standardization")                
```

```python
#linear regression model
```

```python
from sklearn.linear_model import LinearRegression
from  sklearn.metrics import mean_squared_error, r2_score,mean_absolute_error

lr = LinearRegression()
lr.fit(X_train_scaled, y_train)
y_pred = lr.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("R-squared:", r2)
plt.scatter(y_test, y_pred)
```

```python
#lasso regression model
```

```python
from sklearn.linear_model import Lasso
lasso = Lasso()
lasso.fit(X_train_scaled, y_train)
y_pred_lasso = lasso.predict(X_test_scaled)
mae_lasso = mean_absolute_error(y_test, y_pred_lasso)
mse_lasso = mean_squared_error(y_test, y_pred_lasso)
r2_lasso = r2_score(y_test, y_pred_lasso)
print("Mean Absolute Error (Lasso):", mae_lasso)
print("Mean Squared Error (Lasso):", mse_lasso)
print("R-squared (Lasso):", r2_lasso)
plt.scatter(y_test, y_pred_lasso)
```

```python
#cross validation for linear regression lasso
```

```python
from sklearn.linear_model import LassoCV
lasso_cv = LassoCV(cv=5)
lasso_cv.fit(X_train_scaled, y_train)
```

```python
lasso_cv.alpha_
```

```python
y_pred_lasso_cv = lasso_cv.predict(X_test_scaled)
mae_lasso_cv = mean_absolute_error(y_test, y_pred_lasso_cv)
mse_lasso_cv = mean_squared_error(y_test, y_pred_lasso_cv)
r2_lasso_cv = r2_score(y_test, y_pred_lasso_cv)
print("Mean Absolute Error (Lasso CV):", mae_lasso_cv)
print("Mean Squared Error (Lasso CV):", mse_lasso_cv)
print("R-squared (Lasso CV):", r2_lasso_cv)
plt.scatter(y_test, y_pred_lasso_cv)      
```

# ridge

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(X_train_scaled, y_train)
y_pred_ridge = ridge.predict(X_test_scaled)
mae_ridge = mean_absolute_error(y_test, y_pred_ridge)
mse_ridge = mean_squared_error(y_test, y_pred_ridge)
r2_ridge = r2_score(y_test, y_pred_ridge)
print("Mean Absolute Error (Ridge):", mae_ridge)
print("Mean Squared Error (Ridge):", mse_ridge)
print("R-squared (Ridge):", r2_ridge)
plt.scatter(y_test, y_pred_ridge)
```

```python
from sklearn.linear_model import RidgeCV
ridge_cv = RidgeCV(cv=5)
ridge_cv.fit(X_train_scaled, y_train)
y_pred_ridge_cv = ridge_cv.predict(X_test_scaled)
mae_ridge_cv = mean_absolute_error(y_test, y_pred_ridge_cv)
mse_ridge_cv = mean_squared_error(y_test, y_pred_ridge_cv)
r2_ridge_cv = r2_score(y_test, y_pred_ridge_cv)
print("Mean Absolute Error (Ridge CV):", mae_ridge_cv)
print("Mean Squared Error (Ridge CV):", mse_ridge_cv)
print("R-squared (Ridge CV):", r2_ridge_cv)
plt.scatter(y_test, y_pred_ridge_cv)
```
